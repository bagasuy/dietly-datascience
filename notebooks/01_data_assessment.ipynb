## Initial Data Quality Findings

Based on the initial data assessment, several data quality issues were identified:

1. The `date` column is stored as a string and should be converted to a datetime format for time-based analysis.

2. The `health_condition` column contains 490,275 missing values, representing 71.29% of the dataset. The appropriate treatment should be determined based on the role of this feature in the analysis and modeling process.

3. No fully duplicated rows were found in the dataset.

4. The `daily_steps` column contains negative values, with a minimum value of -419. These values are logically invalid and require further investigation during the data cleaning stage.

5. Several categorical variables were identified, including `gender`, `activity_type`, `intensity`, `health_condition`, and `smoking_status`.

6. The dataset contains 687,701 records from 3,000 participants and covers 360 unique dates, indicating a longitudinal structure that can support time-based analysis.

7. The `weight_kg` variable is available and may be relevant to the prediction objective because weight progression is one of the main concerns addressed by Dietly.


```python
# Data Cleaning

df_clean = df.copy()

print("Original shape:", df.shape)
print("Cleaning dataframe shape:", df_clean.shape)
```

```python
# Investigate Health Condition

health_condition_counts = df_clean["health_condition"].value_counts(dropna=False)

display(health_condition_counts.to_frame(name="count"))

print("\nPercentage:")
display(
    (health_condition_counts / len(df_clean) * 100)
    .round(2)
    .to_frame(name="percentage")
)
```

```python
# Remove invalid daily_steps records

before_rows = len(df_clean)

df_clean = df_clean[df_clean["daily_steps"] >= 0].copy()

after_rows = len(df_clean)
removed_rows = before_rows - after_rows

print("Rows before cleaning:", before_rows)
print("Invalid rows removed:", removed_rows)
print("Rows after cleaning:", after_rows)
print("Remaining negative values:", (df_clean["daily_steps"] < 0).sum())
```

```python
# Check Numeric Value Ranges

numeric_columns = [
    "age",
    "height_cm",
    "weight_kg",
    "duration_minutes",
    "calories_burned",
    "avg_heart_rate",
    "hours_sleep",
    "stress_level",
    "daily_steps",
    "hydration_level",
    "bmi",
    "resting_heart_rate",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "fitness_level"
]

range_summary = df_clean[numeric_columns].agg(["min", "max"]).T

display(range_summary)
```

```python
# EDA 5: Correlation of Numerical Variables with Weight

correlation_with_weight = (
    df_clean[numeric_columns]
    .corr()["weight_kg"]
    .sort_values(ascending=False)
)

display(
    correlation_with_weight
    .to_frame(name="correlation_with_weight")
)
```

### EDA Finding: Variables Related to Weight

The correlation analysis shows that several numerical variables have stronger linear relationships with weight. Fitness level has the strongest positive correlation with weight (0.7619), followed by BMI (0.5739), height (0.4728), and calories burned (0.3652). Daily steps show a moderate negative correlation with weight (-0.3440).

In contrast, age, heart rate, hydration, activity duration, sleep duration, stress level, and blood pressure variables show correlations close to zero.

These results provide an initial basis for feature selection. However, correlation with current weight does not necessarily indicate usefulness for predicting future weight. Potential target leakage, particularly for BMI because it is derived from weight and height, should also be considered before model training.

```python
# Check availability of 7-day-ahead observations

df_sorted = df_clean.sort_values(
    ["participant_id", "date"]
).copy()

df_sorted["future_date_7d"] = (
    df_sorted["date"] + pd.Timedelta(days=7)
)

future_7d_check = df_sorted.merge(
    df_sorted[
        ["participant_id", "date", "weight_kg"]
    ],
    left_on=["participant_id", "future_date_7d"],
    right_on=["participant_id", "date"],
    how="left",
    suffixes=("", "_future")
)

available_7d = future_7d_check["weight_kg_future"].notna().sum()
total_records = len(df_sorted)

print("Total records:", total_records)
print("Records with 7-day future weight:", available_7d)
print(
    "Percentage with 7-day future weight:",
    f"{available_7d / total_records * 100:.2f}%"
)
```

```python
# Create 7-day-ahead prediction target

model_df = future_7d_check.copy()

model_df = model_df.rename(
    columns={
        "weight_kg_future": "future_weight_7d"
    }
)

# Keep only records with a valid 7-day future weight
model_df = model_df[
    model_df["future_weight_7d"].notna()
].copy()

print("Modeling dataset shape:", model_df.shape)
print("\nTarget summary:")
display(
    model_df["future_weight_7d"].describe()
)
```

```python
# Inspect available columns for feature selection

print("Available columns:")
for i, column in enumerate(model_df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Check missing values in candidate features

candidate_features = [
    "age",
    "gender",
    "height_cm",
    "weight_kg",
    "activity_type",
    "duration_minutes",
    "intensity",
    "calories_burned",
    "avg_heart_rate",
    "hours_sleep",
    "stress_level",
    "daily_steps",
    "hydration_level",
    "resting_heart_rate",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "health_condition",
    "smoking_status",
    "fitness_level",
]

missing_summary = (
    model_df[candidate_features]
    .isna()
    .sum()
    .to_frame(name="missing_count")
)

missing_summary["missing_percentage"] = (
    missing_summary["missing_count"]
    / len(model_df)
    * 100
).round(2)

display(
    missing_summary.sort_values(
        "missing_percentage",
        ascending=False
    )
)
```

```python
# Define model features

features = [
    "age",
    "gender",
    "height_cm",
    "weight_kg",
    "activity_type",
    "duration_minutes",
    "intensity",
    "calories_burned",
    "avg_heart_rate",
    "hours_sleep",
    "stress_level",
    "daily_steps",
    "hydration_level",
    "resting_heart_rate",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "smoking_status",
    "fitness_level",
]

target = "future_weight_7d"

print("Number of features:", len(features))
print("Target:", target)
```

```python
# Check data types of model features

display(
    model_df[features].dtypes.to_frame(name="data_type")
)

print("\nUnique values for categorical features:")

categorical_features = [
    "gender",
    "activity_type",
    "intensity",
    "smoking_status",
    "fitness_level",
]

for column in categorical_features:
    print(f"\n{column}:")
    print(model_df[column].unique())
```

```python
# Define final candidate features for the first model

numeric_features = [
    "age",
    "height_cm",
    "weight_kg",
    "duration_minutes",
    "calories_burned",
    "avg_heart_rate",
    "hours_sleep",
    "stress_level",
    "daily_steps",
    "hydration_level",
    "resting_heart_rate",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "fitness_level",
]

categorical_features = [
    "gender",
    "activity_type",
    "intensity",
    "smoking_status",
]

features = numeric_features + categorical_features

print("Number of numeric features:", len(numeric_features))
print("Number of categorical features:", len(categorical_features))
print("Total features:", len(features))
print("\nFeatures:")
print(features)
```

```python
# Sort data chronologically
model_df = model_df.sort_values(
    ["participant_id", "date"]
).copy()

# Use the first 80% of dates for training
# and the last 20% for testing
unique_dates = sorted(model_df["date"].unique())

split_index = int(len(unique_dates) * 0.80)
split_date = unique_dates[split_index]

train_df = model_df[
    model_df["date"] < split_date
].copy()

test_df = model_df[
    model_df["date"] >= split_date
].copy()

print("Split date:", split_date)
print("Training data shape:", train_df.shape)
print("Testing data shape:", test_df.shape)

print("\nTraining period:")
print(train_df["date"].min(), "to", train_df["date"].max())

print("\nTesting period:")
print(test_df["date"].min(), "to", test_df["date"].max())
```

```python
# Separate features (X) and target (y)

X_train = train_df[features].copy()
y_train = train_df[target].copy()

X_test = test_df[features].copy()
y_test = test_df[target].copy()

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

print("\nTarget name:", target)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numeric",
            "passthrough",
            numeric_features
        ),
    ]
)

print("Preprocessor created successfully.")
```

```python
# Fit preprocessing on training data only
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the fitted preprocessing to test data
X_test_processed = preprocessor.transform(X_test)

print("Processed X_train shape:", X_train_processed.shape)
print("Processed X_test shape:", X_test_processed.shape)
```

```python
# Check processed data for missing or invalid values

import numpy as np

print("NaN values in X_train:", np.isnan(X_train_processed).sum())
print("NaN values in X_test:", np.isnan(X_test_processed).sum())

print("Infinite values in X_train:", np.isinf(X_train_processed).sum())
print("Infinite values in X_test:", np.isinf(X_test_processed).sum())

print("\nTarget missing values:")
print("y_train:", y_train.isna().sum())
print("y_test:", y_test.isna().sum())
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create baseline model
baseline_model = DummyRegressor(strategy="mean")

# Train baseline
baseline_model.fit(X_train_processed, y_train)

# Predict on test data
baseline_predictions = baseline_model.predict(X_test_processed)

# Evaluate baseline
baseline_mae = mean_absolute_error(y_test, baseline_predictions)
baseline_rmse = np.sqrt(
    mean_squared_error(y_test, baseline_predictions)
)
baseline_r2 = r2_score(y_test, baseline_predictions)

print("Baseline Model Performance")
print("--------------------------")
print(f"MAE:  {baseline_mae:.4f} kg")
print(f"RMSE: {baseline_rmse:.4f} kg")
print(f"R²:   {baseline_r2:.4f}")
```

```python
from sklearn.linear_model import LinearRegression

# Create Linear Regression model
linear_model = LinearRegression()

# Train model
linear_model.fit(X_train_processed, y_train)

# Predict on test data
linear_predictions = linear_model.predict(X_test_processed)

# Evaluate model
linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(
    mean_squared_error(y_test, linear_predictions)
)
linear_r2 = r2_score(y_test, linear_predictions)

print("Linear Regression Performance")
print("-----------------------------")
print(f"MAE:  {linear_mae:.4f} kg")
print(f"RMSE: {linear_rmse:.4f} kg")
print(f"R²:   {linear_r2:.4f}")
```

```python
# Check relationship between current weight and future weight

weight_target_corr = model_df[
    ["weight_kg", "future_weight_7d"]
].corr()

display(weight_target_corr)

print(
    "Correlation between current weight and future weight:",
    round(weight_target_corr.loc["weight_kg", "future_weight_7d"], 4)
)
```

```python
# Baseline using current weight only

X_weight_only = model_df[["weight_kg"]].copy()
y_weight_only = model_df["future_weight_7d"].copy()

# Use the same chronological split
X_weight_train = X_weight_only.loc[train_df.index]
X_weight_test = X_weight_only.loc[test_df.index]

y_weight_train = y_weight_only.loc[train_df.index]
y_weight_test = y_weight_only.loc[test_df.index]

weight_only_model = LinearRegression()

weight_only_model.fit(
    X_weight_train,
    y_weight_train
)

weight_only_predictions = weight_only_model.predict(X_weight_test)

weight_only_mae = mean_absolute_error(
    y_weight_test,
    weight_only_predictions
)

weight_only_rmse = np.sqrt(
    mean_squared_error(y_weight_test, weight_only_predictions)
)

weight_only_r2 = r2_score(
    y_weight_test,
    weight_only_predictions
)

print("Weight-Only Linear Regression")
print("-----------------------------")
print(f"MAE:  {weight_only_mae:.4f} kg")
print(f"RMSE: {weight_only_rmse:.4f} kg")
print(f"R²:   {weight_only_r2:.4f}")
```

```python
# Model without current weight and BMI

features_without_weight = [
    "age",
    "height_cm",
    "duration_minutes",
    "calories_burned",
    "avg_heart_rate",
    "hours_sleep",
    "stress_level",
    "daily_steps",
    "hydration_level",
    "resting_heart_rate",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "fitness_level",
    "gender",
    "activity_type",
    "intensity",
    "smoking_status",
]

X_train_no_weight = train_df[features_without_weight].copy()
X_test_no_weight = test_df[features_without_weight].copy()

y_train_no_weight = train_df[target].copy()
y_test_no_weight = test_df[target].copy()

# Separate categorical and numerical features
categorical_no_weight = [
    "gender",
    "activity_type",
    "intensity",
    "smoking_status",
]

numeric_no_weight = [
    feature
    for feature in features_without_weight
    if feature not in categorical_no_weight
]

# Preprocessing
preprocessor_no_weight = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_no_weight,
        ),
        (
            "numeric",
            "passthrough",
            numeric_no_weight,
        ),
    ]
)

# Transform data
X_train_no_weight_processed = (
    preprocessor_no_weight.fit_transform(X_train_no_weight)
)

X_test_no_weight_processed = (
    preprocessor_no_weight.transform(X_test_no_weight)
)

print(
    "Processed X_train shape:",
    X_train_no_weight_processed.shape
)

print(
    "Processed X_test shape:",
    X_test_no_weight_processed.shape
)
```

```python
# Train Linear Regression without current weight and BMI

linear_no_weight_model = LinearRegression()

linear_no_weight_model.fit(
    X_train_no_weight_processed,
    y_train_no_weight
)

# Predict
linear_no_weight_predictions = (
    linear_no_weight_model.predict(
        X_test_no_weight_processed
    )
)

# Evaluate
linear_no_weight_mae = mean_absolute_error(
    y_test_no_weight,
    linear_no_weight_predictions
)

linear_no_weight_rmse = np.sqrt(
    mean_squared_error(
        y_test_no_weight,
        linear_no_weight_predictions
    )
)

linear_no_weight_r2 = r2_score(
    y_test_no_weight,
    linear_no_weight_predictions
)

print("Linear Regression Without Weight & BMI")
print("---------------------------------------")
print(f"MAE:  {linear_no_weight_mae:.4f} kg")
print(f"RMSE: {linear_no_weight_rmse:.4f} kg")
print(f"R²:   {linear_no_weight_r2:.4f}")
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Participant-level train/test split
group_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    group_split.split(
        model_df,
        groups=model_df["participant_id"]
    )
)

train_df_group = model_df.iloc[train_idx].copy()
test_df_group = model_df.iloc[test_idx].copy()

print("Train rows:", len(train_df_group))
print("Test rows:", len(test_df_group))

print(
    "Train participants:",
    train_df_group["participant_id"].nunique()
)

print(
    "Test participants:",
    test_df_group["participant_id"].nunique()
)

participant_overlap = set(
    train_df_group["participant_id"]
).intersection(
    set(test_df_group["participant_id"])
)

print(
    "Overlapping participants:",
    len(participant_overlap)
)

if len(participant_overlap) == 0:
    print("PASS: No participant overlap between train and test.")
else:
    print("WARNING: Participant overlap detected.")
```

```python
# Define features and target for participant-level split

target = "future_weight_7d"

features = [
    "weight_kg",
    "height_cm",
    "age",
    "fitness_level",
    "calories_burned",
    "daily_steps",
    "resting_heart_rate",
    "hours_sleep",
    "hydration_level",
    "duration_minutes",
    "stress_level",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "avg_heart_rate",
    "intensity",
    "smoking_status",
]

X_train_group = train_df_group[features].copy()
y_train_group = train_df_group[target].copy()

X_test_group = test_df_group[features].copy()
y_test_group = test_df_group[target].copy()

print("X_train shape:", X_train_group.shape)
print("X_test shape:", X_test_group.shape)
print("y_train shape:", y_train_group.shape)
print("y_test shape:", y_test_group.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Separate numeric and categorical features

numeric_features = [
    "weight_kg",
    "height_cm",
    "age",
    "fitness_level",
    "calories_burned",
    "daily_steps",
    "resting_heart_rate",
    "hours_sleep",
    "hydration_level",
    "duration_minutes",
    "stress_level",
    "blood_pressure_systolic",
    "blood_pressure_diastolic",
    "avg_heart_rate",
]

categorical_features = [
    "intensity",
    "smoking_status",
]

preprocessor_group = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            numeric_features,
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False,
            ),
            categorical_features,
        ),
    ]
)

# Fit ONLY on training data
X_train_group_processed = preprocessor_group.fit_transform(
    X_train_group
)

# Transform test data using the fitted training preprocessor
X_test_group_processed = preprocessor_group.transform(
    X_test_group
)

print(
    "X_train_group_processed shape:",
    X_train_group_processed.shape
)

print(
    "X_test_group_processed shape:",
    X_test_group_processed.shape
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# =========================
# Linear Regression
# =========================

linear_group_model = LinearRegression()

linear_group_model.fit(
    X_train_group_processed,
    y_train_group
)

linear_group_predictions = linear_group_model.predict(
    X_test_group_processed
)

linear_group_mae = mean_absolute_error(
    y_test_group,
    linear_group_predictions
)

linear_group_rmse = np.sqrt(
    mean_squared_error(
        y_test_group,
        linear_group_predictions
    )
)

linear_group_r2 = r2_score(
    y_test_group,
    linear_group_predictions
)

print("Participant-Level Linear Regression")
print("-----------------------------------")
print(f"MAE:  {linear_group_mae:.4f} kg")
print(f"RMSE: {linear_group_rmse:.4f} kg")
print(f"R²:   {linear_group_r2:.4f}")
```

```python
# =========================
# Participant-Level Random Forest
# =========================

random_forest_group_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest_group_model.fit(
    X_train_group_processed,
    y_train_group
)

random_forest_group_predictions = random_forest_group_model.predict(
    X_test_group_processed
)

random_forest_group_mae = mean_absolute_error(
    y_test_group,
    random_forest_group_predictions
)

random_forest_group_rmse = np.sqrt(
    mean_squared_error(
        y_test_group,
        random_forest_group_predictions
    )
)

random_forest_group_r2 = r2_score(
    y_test_group,
    random_forest_group_predictions
)

print("Participant-Level Random Forest")
print("--------------------------------")
print(f"MAE:  {random_forest_group_mae:.4f} kg")
print(f"RMSE: {random_forest_group_rmse:.4f} kg")
print(f"R²:   {random_forest_group_r2:.4f}")
```

```python
from sklearn.pipeline import Pipeline

# ==========================================
# Final Pipeline: Preprocessing + Regression
# ==========================================

final_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor_group),
        ("model", LinearRegression())
    ]
)

# Train only on participant-level training data
final_pipeline.fit(
    X_train_group,
    y_train_group
)

# Predict on participant-level test data
final_pipeline_predictions = final_pipeline.predict(
    X_test_group
)

# Evaluate final pipeline
final_pipeline_mae = mean_absolute_error(
    y_test_group,
    final_pipeline_predictions
)

final_pipeline_rmse = np.sqrt(
    mean_squared_error(
        y_test_group,
        final_pipeline_predictions
    )
)

final_pipeline_r2 = r2_score(
    y_test_group,
    final_pipeline_predictions
)

print("Final Linear Regression Pipeline")
print("---------------------------------")
print(f"MAE:  {final_pipeline_mae:.4f} kg")
print(f"RMSE: {final_pipeline_rmse:.4f} kg")
print(f"R²:   {final_pipeline_r2:.4f}")
```
